```python
import pandas as pd
import numpy as np

# load sheet 1 for campaign records
df = pd.read_excel('Crowdfunding campaign.xlsx', sheet_name='Sheet1')
print("First 5 rows:")
print(df.head())

# Check how many total rows and columns exist before cleaning
print("\nOriginal dataset shape (rows, columns):", df.shape)
```

```python
# checking for duplicate rows
num_duplicates = df.duplicated().sum()
print(f"Number of duplicate rows: {num_duplicates}")

# remove duplicate rows
df_clean = df.drop_duplicates().copy()

# reset index to fix row numbers since rows were deleted
df_clean.reset_index(drop=True, inplace=True)

# new dataset shape
print("Shape after removing duplicates:", df_clean.shape)
```

```python
# look for missing vals
print("Missing values per column:")
print(df_clean.isnull().sum())
```

```python
# delivery has 171 missing vals since it happens only after a campaign is finished
# if we were to delete rows with null delivery, the dataset would lose 7% of its data, so we will just drop the column
df_clean = df_clean.drop(columns=['delivery'])
print("Columns remaining:", df_clean.columns.tolist())
```

```python
# drop rows that have missing vals in fixed_fund and iot_plus
df_clean = df_clean.dropna(subset=['fixed_fund', 'iot_plus'])

# binary columns need to be stored as integers
binary_cols = ['pre_order_perk', 'tech_campaign', 'sbiz_campaign', 'fixed_fund', 'iot_plus']
df_clean[binary_cols] = df_clean[binary_cols].astype(int)

# check and print new shape + make sure no missing vals remain
print("Final shape after handling missing values:", df_clean.shape)
print("\nRemaining missing values:")
print(df_clean.isnull().sum())
```

```python
# look at summary stats for key numbers
print(df_clean[['amt_goal', 'frac_raised']].describe())

# check what years campaigns are from
print("\nCampaign counts by start year:")
print(df_clean['year_start'].value_counts())
```

```python
# save cleaned dataset to a new csv file
df_clean.to_csv('cleasned_crowdfunding_data.csv', index=False)
print("Cleaned data successfully saved to 'cleaned_crowdfunding_data.csv'!")
```
